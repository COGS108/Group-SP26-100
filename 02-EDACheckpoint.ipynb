# Research Question

**Definitions**
 - **NBA player workload** means pre-game summaries we build from schedules and box scores (recent minutes, games in short calendar windows, rest days, back-to-back or dense schedules, travel). 

 - **Injury outcome** means after each reference game, the player either has a game-reported injury tied to their next scheduled game (injury report for that game and/or did not play in that next game).

**Question:** Does increasing NBA player workload before a reference game increase the chance of a next-game injury outcome, and which workload variable has the strongest association with that outcome, measured by the largest absolute correlation and/or the largest standardized coefficient in baseline regression models?

## Background and Prior Work

NBA injuries are important to study because they affect player health and player availability. When players get hurt, they may miss games or require changes to team decisions about player usage. Mack et al. analyzed NBA injuries from the 2013-2014 through 2018-2019 seasons and found between 1,550 and 1,892 injuries per season, with 33.6% to 38.5% of those injuries resulting in at least one missed NBA game. Their study also reported game-loss injury rates per 10,000 player-minutes, which makes minutes played a relevant workload variable for studying injury risk.<a name="cite_ref-bg-1"></a>[<sup>1</sup>](#cite_note-bg-1)

Our research question focuses on workload because NBA players go through a long regular season with frequent games, high minutes, travel, and limited recovery time. Workload can include how many minutes a player plays, how often they play, how many rest days they get, and how far they travel between games. The NBA has also recognized that schedule congestion can affect player recovery. For the 2017-2018 season, the league extended the regular-season schedule window, eliminated four-games-in-five-nights stretches, and reduced back-to-back games from 16.3 to 14.4 per team.<a name="cite_ref-bg-2"></a>[<sup>2</sup>](#cite_note-bg-2) This supports studying rest days, game frequency, minutes played, and travel distance as possible predictors of injury.

Previous NBA research shows that the connection between schedule difficulty and injuries is not simple. Teramoto et al. studied NBA game injuries from the 2012-2013 through 2014-2015 regular seasons and examined schedule factors such as back-to-back games, four games in five days, and away games. They found that playing back-to-back games or four games in five days alone was not associated with a higher rate of game injuries, but playing away from home was positively associated with game injuries. They also found that back-to-back games and away games were significant predictors of players having frequent game injuries.<a name="cite_ref-bg-3"></a>[<sup>3</sup>](#cite_note-bg-3) This matters for our project because it suggests we should compare several workload features together instead of assuming that one factor, such as back-to-backs, explains injury risk by itself.

Other basketball research suggests that recent workload can be studied relative to what a player is normally used to. Weiss et al. studied professional basketball players using the acute:chronic workload ratio, which compares a player’s current weekly workload to their average workload from the previous four weeks. They reported that fewer team members were injured after workload ratios between 1.0 and 1.49 than after very low, low, or high workload ratios, although they also noted that the mean differences in injury likelihood were unclear and that workload-injury trends varied by player.<a name="cite_ref-bg-4"></a>[<sup>4</sup>](#cite_note-bg-4) This supports using recent workload features, such as minutes played in the last five games or games played in the last 7, 14, or 30 days, while still being careful not to treat workload as a simple one-variable cause of injury.

Travel is also relevant because it can affect sleep, fatigue, and recovery. Charest et al. studied NBA back-to-back games from 2013-2020 and looked at travel distance, travel direction, and game order. Their study focused on winning percentage rather than injuries, but they found that travel distance and travel patterns were related to performance outcomes. They also explained that travel fatigue and circadian disruption can disturb sleep and recovery across an NBA season.<a name="cite_ref-bg-5"></a>[<sup>5</sup>](#cite_note-bg-5) For our project, this supports including travel distance as a possible injury-risk feature, while being careful not to claim that travel distance directly causes injuries unless our own analysis supports that conclusion.

An especially similar prior project is Cohan et al.’s NBA injury prediction model. They used public injury data, player-game data, game activity, and player statistics to predict future injuries. Their dataset included about 238,000 player-game observations, but only about 5,000 injuries were associated with games, meaning injuries were rare compared with non-injury observations. They also found that high recent playing-time load, defined as more than 100 minutes in the previous five games, influenced injury incidence.<a name="cite_ref-bg-6"></a>[<sup>6</sup>](#cite_note-bg-6) Our project builds on this idea by focusing on easy-to-understand workload features, such as recent minutes, rest days, game frequency, and travel distance, and testing which of those variables contribute most to higher injury risk.

# Hypothesis


**Definitions**
 - **NBA player workload** means the pre-game workload summaries we build from schedules and box scores, including recent minutes played, number of games in short calendar windows, rest days, back-to-back or dense-schedule indicators, and travel.

 - **Injury outcome** means a binary next-game label: after each reference game, the player either has a game-reported injury tied to their next scheduled game and/or does not play in that next game because of injury.

**Hypothesis:** Based on the background research, we predict a strong positive correlation between higher NBA player workload and the next-game injury outcome. This means that as a player's recent minutes, schedule density, travel burden, or lack of rest increases before a game, their chance of an injury-related absence or injury report entry in the next scheduled game will also increase.

We also expect the strength of this relationship to vary across workload measures. Since prior research suggests schedule effects are not simple, we do not assume that every workload variable will matter equally. We expect recent minutes played and travel/rest-related workload measures to show the strongest associations with the injury outcome, measured by absolute correlation values and/or standardized regression coefficients.

## Data

### Data overview

- **Dataset #1: NBA Injury Stats (1951-2023)**
  - Link to the dataset: https://www.kaggle.com/datasets/loganlauton/nba-injury-stats-1951-2023
  
    The dataset has 37667 observations and 5 columns: Date, Team, Acquired, Relinquished, and Notes. All 5 columns start as pandas objects (strings), but Date--which gives the date a player is placed or taken off the IL (Injured List)--will be cast to datetime64 so we can filter by season and measure injury lengths. The rest of the columns will stay as objects as they contain textual information. Acquired and Relinquished show exactly which player is removed from or placed onto the injury report list, respectively. Along with the date variable, which records the date of the roster transaction, Acquired and Relinquished allow us to calculate the number of games a player has missed over the course of a season and the playoffs. It's important to note that there are many null values in both the Acquired and Relinquished columns. This is because the columns are complementary--if a player is coming off the IL, then they are not being added, and vice versa. The Team variable records the team of the player moving to or from the IL. The Notes variable contains free-text that lists what types of injuries or issues a player faces that lead them to be on the injury report list.
    
    One concern about the dataset is that since Notes contains unstructured free text, it does not show or list all injuries or reasons why a player is placed onto the injury report list, with many observations being inconsistently worded on the types of injuries a player may have faced. In other words, Notes only contain the team's official injury report of the player, not a full description of their health status. Minor injuries that players play through or that teams handle without an IL move (game-time minutes restriction or sitting out) are not recorded in the dataset. Another concern that comes with unstructured free text is that we have to parse through the strings to filter out non-injury reasons for leave (ex. Military Leave). The data set also does not come with the number of games a player misses due to being on the injury report, so we would have to calculate that using the Date column.
    
    The dataset should also be more focused on the last 20 years, as it more closely represents the current state and pace of how NBA games are played. This also helps fix the problem of removing teams that are no longer in the NBA, like the Bullets.


- **Dataset #2: NBA Game Performance and Team Statistics (2004 - 2022)**
  - Dataset Link: https://www.kaggle.com/datasets/nathanlauga/nba-games

    The second dataset provides an extensive collection of game results, team rankings, and individual player box scores spanning from the 2004 season through 2022. It is a multi-relational dataset consisting of four primary files: games.csv (25,261 observations), games_details.csv (645,963 observations), players.csv (1,749 observations), and teams.csv (30 observations). While games.csv tracks high-level outcomes such as total points (PTS) and team-wide shooting percentages (FG_PCT, FT_PCT), the games_details.csv file provides granular, player-level metrics for every matchup. Key variables include Points (PTS), Assists (AST), and PLUS_MINUS—a raw impact metric representing the point differential while a specific player is on the court. To make this data actionable for longitudinal study, the player and team IDs must be mapped across files to link specific performances to franchise metadata.

    A primary concern regarding this dataset is the presence of "Self-Selection" or Contextual Bias. Because the observations consist of professional athletes in a highly controlled environment, these individuals are statistical outliers, and their performance may be skewed by "garbage time"—minutes played at the end of a blowout game that do not reflect true competitive skill. Furthermore, the dataset is limited to traditional box score statistics. It lacks modern advanced metrics such as Player Efficiency Rating (PER), True Shooting Percentage (TS%), or tracking-based usage rates, which are often necessary to assess a player's actual efficiency.

    Another significant limitation involves "Temporal Shift." The NBA has undergone a "Three-Point Revolution" since 2015; consequently, shooting data and offensive distributions from 2004 look fundamentally different from those in 2022. Without normalization, direct comparisons across these different eras may lead to misleading conclusions about a player's relative impact. Finally, while the data is highly granular at the per-game level, it requires significant computational aggregation to identify the season-long trends or career trajectories necessary for predictive modeling.

        - vorp (value over replacement): The value of a player over a replacement calliber player in the league.
        - usg_percent: The percentage of time a player has possesion of the ball while they are out on the court.
        - bpm:  The point differential when a specific player is on the court. It is a raw unit of impact, where a positive value indicates the team outscored the opponent during that player's minutes.

        - Self-Selection/Contextual Bias: As stated before data and stats like usg_percent can be heavily influenced by factors such as 'garbage time' as well as teams intentionally tankning.

### Dataset #3 — NBA Advanced Stats (2000 - 2025)

Important Metrics:

WS: Winshares is an advanced metric calculating the amount of wins out of 82 games that a player themselves contributes to a team. This advanced stat is calculated through how well a player performs on both the offensive side of the court as well as the defensive side of the court. In this stat players with 0-2 win shares are usually considered bench or replacement level players having a minimal role on the court or rarely playing at all. At the 2-5 winshare range, players are typically considered rotational/role players. These players usually average around 15-30 minutes a game with expertise in a few specific skills that help support star players. At 5-8 winshares, players are usually starters on their teams, playing the most minutes and are typically a teams best five man rotation. At 8-12 winshares, players are typically in contention for All-star Selections.These players are typically the main or second contributor to a team's success whether it be offensively or defensively. At 12-18+ winshares players are usually in contention for the MVP award. These players are the best and most impactful players on a winning team, often having the greatest workload on their team.

VORP: Value over replacement is an advanced metric derived from the box score in basketball that quantifies a player's total contribution to their team compared to a theoretical "replacement level" player. Having a VORP of 0.0 makes you the average NBA player. Having a VORP of 2.0-4.0 makes you a solid starter. At the range of 5.0-7.0, players are usually at an All-Star level.  If a player has a VORP of 7.0, they are usually an MVP level canidate. 

usg_percent: Usage percent or how often the ball is in a players possession while they are on the floor. This can be highly misleading as players with minimal minutes but get more touches based off coincidence can lead to having a higher usage percentage compared to players that play meaningful minutes.

#### Minutes Played by Next-Game Injury Outcome

This boxplot compares minutes played for player-games that were or were not followed by an injury-list placement before the team's next scheduled game. EDA also identifies the associations rather than causual relationships between injury and minutes played due to possible confounding variables.

```python
import os
os.environ.setdefault('XDG_CACHE_HOME', '/tmp/cogs108-cache')
os.environ.setdefault('MPLCONFIGDIR', '/tmp/cogs108-matplotlib')
os.makedirs(os.environ['XDG_CACHE_HOME'], exist_ok=True)
os.makedirs(os.environ['MPLCONFIGDIR'], exist_ok=True)

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import re
import unicodedata
from IPython.display import display


def normalize_player_name(value):
    if pd.isna(value):
        return np.nan
    value = unicodedata.normalize('NFKD', str(value)).encode('ascii', 'ignore').decode('ascii')
    value = value.lower()
    value = re.sub(r'\([^)]*\)', '', value)
    value = re.sub(r'\b(jr|sr|ii|iii|iv|v)\b\.?', '', value)
    value = re.sub(r'[^a-z ]+', ' ', value)
    return re.sub(r'\s+', ' ', value).strip()


def minutes_to_float(value):
    if pd.isna(value):
        return np.nan
    parts = str(value).split(':')
    if len(parts) == 2:
        return float(parts[0]) + float(parts[1]) / 60
    return pd.to_numeric(value, errors='coerce')


games_eda = pd.read_csv('data/02-processed/games_clean.csv')
details_eda = pd.read_csv('data/02-processed/games_details.csv', low_memory=False)
injuries_eda = pd.read_csv('data/02-processed/nba_injury_stats_clean.csv')
games_eda['GAME_DATE_EST'] = pd.to_datetime(games_eda['GAME_DATE_EST'])
injuries_eda['Date'] = pd.to_datetime(injuries_eda['Date'])

injury_start = injuries_eda['Date'].min()
injury_end = injuries_eda['Date'].max()
games_eda = games_eda[
    games_eda['GAME_DATE_EST'].between(injury_start, injury_end)
].copy()

team_games = pd.concat([
    games_eda[['GAME_ID', 'GAME_DATE_EST', 'SEASON', 'HOME_TEAM_ID']].rename(columns={'HOME_TEAM_ID': 'TEAM_ID'}),
    games_eda[['GAME_ID', 'GAME_DATE_EST', 'SEASON', 'VISITOR_TEAM_ID']].rename(columns={'VISITOR_TEAM_ID': 'TEAM_ID'})
], ignore_index=True).drop_duplicates()
team_games = team_games.sort_values(['TEAM_ID', 'SEASON', 'GAME_DATE_EST', 'GAME_ID'])
team_games['next_team_game_date'] = team_games.groupby(['TEAM_ID', 'SEASON'])['GAME_DATE_EST'].shift(-1)

player_games = details_eda.merge(
    team_games[[
        'GAME_ID', 'TEAM_ID', 'GAME_DATE_EST', 'SEASON', 'next_team_game_date'
    ]],
    on=['GAME_ID', 'TEAM_ID'],
    how='inner'
)
player_games['minutes_played'] = player_games['MIN'].apply(minutes_to_float)
player_games = player_games[
    player_games['minutes_played'].notna()
    & (player_games['minutes_played'] > 0)
    & player_games['next_team_game_date'].notna()
].copy()
player_games['player_key'] = player_games['PLAYER_NAME'].apply(normalize_player_name)

injury_events = injuries_eda[injuries_eda['Relinquished'].notna()].copy()
injury_events['player_key'] = injury_events['Relinquished'].apply(normalize_player_name)
injury_events = injury_events[['Date', 'player_key']].drop_duplicates()

label_rows = player_games[
    ['PLAYER_ID', 'player_key', 'GAME_ID', 'GAME_DATE_EST', 'next_team_game_date']
].merge(injury_events, on='player_key', how='left')
label_rows['injury_before_next_team_game'] = (
    label_rows['Date'].notna()
    & (label_rows['Date'] > label_rows['GAME_DATE_EST'])
    & (label_rows['Date'] <= label_rows['next_team_game_date'])
)
injury_labels = label_rows.groupby(
    ['PLAYER_ID', 'GAME_ID'], as_index=False
)['injury_before_next_team_game'].any()

eda_df = player_games.merge(injury_labels, on=['PLAYER_ID', 'GAME_ID'], how='left')
eda_df['injury_before_next_team_game'] = (
    eda_df['injury_before_next_team_game'].fillna(False).astype(int)
)

outcome_order = ['No next-game IL placement', 'Next-game IL placement']
eda_df['injury_outcome'] = pd.Categorical(
    eda_df['injury_before_next_team_game'].map({0: outcome_order[0], 1: outcome_order[1]}),
    categories=outcome_order,
    ordered=True
)

workload_summary = (
    eda_df.groupby('injury_outcome', observed=True)
    .agg(
        player_games=('PLAYER_ID', 'size'),
        median_minutes=('minutes_played', 'median'),
        mean_minutes=('minutes_played', 'mean')
    )
    .round(2)
)
display(workload_summary)

minutes_groups = [
    eda_df.loc[eda_df['injury_before_next_team_game'].eq(0), 'minutes_played'],
    eda_df.loc[eda_df['injury_before_next_team_game'].eq(1), 'minutes_played']
]
plot_labels = ['No next-game\nIL placement', 'Next-game\nIL placement']

fig, ax = plt.subplots(figsize=(7, 5))
box = ax.boxplot(minutes_groups, labels=plot_labels, showfliers=False, patch_artist=True)
for patch, color in zip(box['boxes'], ['#9ecae1', '#fdae6b']):
    patch.set_facecolor(color)
for median in box['medians']:
    median.set_color('black')
ax.set_title('Reference-game minutes')
ax.set_xlabel('Next-game injury-list outcome')
ax.set_ylabel('Minutes played')
ax.grid(axis='y', alpha=0.3)

fig.tight_layout()
plt.show()
```

#### Interpretation of Minutes Played


The boxplot above shows that player-games followed by next-game injury list placement had lower minutes in their reference game than the player-games not followed by injury list placement. Non-injury group showing a median of 23.47 minutes with a mean of 23.01 minutes, while the injury group shows lower over all play time. The injury group has a median of 17.33 minutes and a mean of 18.21 minutes.

This result does not support the positive workload hypothesis if workload is measured only by minutes played in the reference game. Player-games followed by injury-list placement had lower reference-game minutes, which may reflect players already being limited by soreness, fatigue, injury management, or other contextual factors before they were formally placed on the injury list.

The plot suggests that single-game minutes may not be sufficient in injury prediction and workload-related risks. Rolling workload intensity, schedule density, travel burden, and rest days could better refine predictions of injury-risk. Because the EDA is observational, there is no definite conclusion that workload is a direct cause of reduced minutes/injury outcome from the plot alone. However, it supports the neccessity of cumulative examinations with other related variables without relaince soley on minutes played as a determining factor.

#### Demographics of High-Risk, High-Workload Players

We isolate danger-zone player-games where a player averaged ≥36 minutes over their last 5 games**, then show how that heavy-workload pool splits by years in the league (0–3 = Young/Fresh, 4–8 = Prime, 9+ = Veteran). This highlights whether age/experience interacts with workload, useful for deciding if terms like Age × Recent Minutes belong in later models.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from IPython.display import display

DANGER_ZONE_MIN_AVG = 36 

EXPERIENCE_BRACKETS = [
    ('Young / Fresh (0–3 yrs)', 0, 3),
    ('Prime (4–8 yrs)', 4, 8),
    ('Veteran (9+ yrs)', 9, np.inf),
]
BRACKET_ORDER = [label for label, _, _ in EXPERIENCE_BRACKETS]
BRACKET_COLORS = ['#74c476', '#6baed6', '#fd8d3c']


def minutes_to_float(value):
    if pd.isna(value):
        return np.nan
    parts = str(value).split(':')
    if len(parts) == 2:
        return float(parts[0]) + float(parts[1]) / 60
    return pd.to_numeric(value, errors='coerce')


def years_in_league_bracket(years):
    for label, low, high in EXPERIENCE_BRACKETS:
        if low <= years <= high:
            return label
    return np.nan


games_eda = pd.read_csv('data/02-processed/games_processed.csv')
details_eda = pd.read_csv('data/02-processed/details_processed.csv', low_memory=False)

games_eda['GAME_DATE_EST'] = pd.to_datetime(games_eda['GAME_DATE_EST'])

team_games = pd.concat([
    games_eda[['GAME_ID', 'GAME_DATE_EST', 'SEASON', 'HOME_TEAM_ID']].rename(
        columns={'HOME_TEAM_ID': 'TEAM_ID'}
    ),
    games_eda[['GAME_ID', 'GAME_DATE_EST', 'SEASON', 'VISITOR_TEAM_ID']].rename(
        columns={'VISITOR_TEAM_ID': 'TEAM_ID'}
    ),
], ignore_index=True).drop_duplicates()

player_games = details_eda.merge(
    team_games[['GAME_ID', 'TEAM_ID', 'GAME_DATE_EST', 'SEASON']],
    on=['GAME_ID', 'TEAM_ID'],
    how='inner',
)
player_games['minutes_played'] = player_games['MIN'].apply(minutes_to_float)
player_games = player_games[
    player_games['minutes_played'].notna() & (player_games['minutes_played'] > 0)
].copy()

player_games = player_games.sort_values(
    ['PLAYER_ID', 'GAME_DATE_EST', 'GAME_ID']
)
player_games['minutes_last_5_avg'] = (
    player_games.groupby('PLAYER_ID')['minutes_played']
    .transform(lambda s: s.rolling(5, min_periods=5).mean())
)

player_games['first_season_in_data'] = player_games.groupby('PLAYER_ID')['SEASON'].transform('min')
player_games['years_in_league'] = player_games['SEASON'] - player_games['first_season_in_data']
player_games['experience_bracket'] = player_games['years_in_league'].apply(years_in_league_bracket)

danger_zone = player_games[
    player_games['minutes_last_5_avg'].ge(DANGER_ZONE_MIN_AVG)
    & player_games['experience_bracket'].notna()
].copy()

bracket_counts = (
    danger_zone['experience_bracket']
    .value_counts()
    .reindex(BRACKET_ORDER, fill_value=0)
)
bracket_pcts = (bracket_counts / bracket_counts.sum() * 100).round(2)

demographics_summary = pd.DataFrame({
    'danger_zone_player_games': bracket_counts,
    'percent_of_pool': bracket_pcts,
})
display(demographics_summary)

fig, ax = plt.subplots(figsize=(8, 8))
wedges, texts, autotexts = ax.pie(
    bracket_counts,
    labels=bracket_counts.index,
    autopct=lambda pct: f'{pct:.1f}%\n(n={int(round(pct / 100 * bracket_counts.sum())):,})',
    colors=BRACKET_COLORS,
    startangle=90,
    textprops={'fontsize': 10},
)
for autotext in autotexts:
    autotext.set_color('black')
ax.set_title(
    'Experience mix in the heavy-workload “danger zone”\n'
    f'(≥{DANGER_ZONE_MIN_AVG} min/game avg over last 5 games, n={bracket_counts.sum():,})'
)
ax.axis('equal')
fig.tight_layout()
plt.show()
```

#### Danger Zone Interpretation

This section examines whether heavy recent workload is concentrated in particular experience groups.

The pie chart shows the experience breakdown of players within heavy-workload "danger-zone" which is defined as a time where the average player time is 36 minutes per game over the course of 5 games. The largest share of the high workload group is comprised of young/fresh players with 0-3 years of experience within the NBA with a total of 27,485 games/ 54.23% of the danger-zone. Prime players have 4-8 years of experience and make of the second largest share with 19,436 games / 38.35% of the pie-chart. Veterans with 9+ years of experience are the smallest by far with only 3,761 games / 7.42%.

The pie chart suggests that recent heavy workloads are common amongst younger and prime career players are much higher than veteran players within the league. It can be explained that the heavier minute play times are given to younger players due to their position as emerging starters or key rotation players. Veteran players may receive more rest, load management, or reduced minutes in comparison for the preservation of health and lower injury-risk. Prime players as established contributors to team formation are also in the danger zone.

This EDA supports an idea involved within our hypothesis, that workload risk is not evenly distributed across player experience levels. If the connection between heavy recent workload and injury is positive, younger/prime players would face increased risk due to heavier playtime exposure. #### Interpretation of Univariate Distributions of Workload Predictors

The above graphs show the univariate player games distributions of different workload predictors of the data. These summarize four common NBA workload variables that are necessary as predictors of injury risk. 

Minutes played(single game): The following distribution represents the total minutes played each player in a single game. The distribution is roughly bell-shaped curved centered around the median of 23.4 minutes played in a game. The single game minutes distribution is a proxy for acute workload. Most players tend to fall within the 15-35 minute range with players above that with high workloads being relatively uncommon. However, these high workload games could be a predictor when including games with short rest and cumulative workload. There are also many lower minute appearances caused by bench players, garbage time, and injury recovery. This creates the lower miunute tail seen in the graph.

Avg minutes, last 5 games: The distribution represents the average minutes players played over the last 5 games with a median of 23.5 minutes. The distribution captures chronic or rolling workload instead of single game exposure, and reduces the variablity and ouliers. This distribution allows better analysis for sustained fatigue and cumulative stress that NBA players face over a longer stretch of time. As a result there is a much strong seperation between bench and starter workloads along with establishing players rotational patterns when analyzing player workload.

Rest days since previous game: The following distribution is a right-skewed distribution with a median of 2. This graph depicts the amount of days of rest a player typically gets before having to play another game. The graph shows how most NBA games typically occur with 1-3 days of rest with longer breaks being caused by injury absences, schedule gaps, or the All-Star break. On days with low rest, players experience increased fatigue and have reduced tissue recovery time. On the flip side, high rest generally aids recovery, however long layoffs can lead to condition loss and potential reinjury after return.

Games played in prior 7 days: The following distribution is a discrete workload distribution of the amount of games each player plays within a week. With a median of 3 games, most players typicallyh in appear in about 2-4 games per week, with dense schedules of 5+ games being rare. This is a strong marker of how condensed a players schedule may be leading to players dealing with things such as cumulative fatigue and travel burden. Higher values in this distribution results in players having lower recovery capacity along with reduced athleticisim during games. This a great predictor as two players with the same amount of minutes averaged may face very different schedules. For example, a player playing 5 games in a week would have greater cumulative fatigue in comparisson to the player who played 2 games in a week. Despite having the same averages in minutes, the cumulative fatigue from widely different workloads results in different risk levels for injury.
